```python
#indexing
number=(20,30,40,50,60)
print(number[2])
print(number[3])
print(number[-3])
```

```python
#slicing-accesing a particular portion
number=(20,30,40,50,60)
print(number[1:4])
print(number[:2])
print(number[2:])
print(number[::-1])
```

```python
#tuple immutablility- we cannot add, remove elements from an existing tuple we got error tuple doesnot support item assignment
numbers=(10,20,30,40,50)
numbers[1]=200
print(numbers)
```

```python
numbers=(100,200,300,400,500) #we didnot modify existing value but we created new tuple and assign then a value
numbers = (600,) + numbers[1:]
print(numbers)
```

```python
#tuples have only 2 in built methods
#count= how many times an value appear in tuple
a=(20,30,45,10,20,22,20)
result=a.count(20)
print(result)
```

```python
#membership operator in and not in
name=("Arya","Priya","Vrushali","Shreeja")
print("Rahul" in name)
print("Vrushali" in name)
```

```python
#looping in tuples= accessing elements with for and while loop
sales=(1000,2000,2500,3000,5000)
for i in sales:
      print(i)
```

```python
#enumerate() gives you both the index and the element.
names=("Amit", "Priya", "Rahul")
for index,name in enumerate(names):
    print(index,name)
```
